### Data Analysis and Visualization


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style
sns.set_theme(style="whitegrid")

# 1. Distribution of Area Types
plt.figure(figsize=(6, 4))
sns.countplot(data=df_cleaned, x='area_type', palette='viridis')
plt.title('Distribution of Area Types')
plt.xlabel('Area Type')
plt.ylabel('Count')
plt.show()
```

```python
# 2. Distribution of Continents (excluding 'Not Applicable')
plt.figure(figsize=(10, 5))
continent_data = df_cleaned[df_cleaned['continent'] != 'Not Applicable']
sns.countplot(data=continent_data, y='continent', order=continent_data['continent'].value_counts().index, palette='muted')
plt.title('Number of Entries per Continent')
plt.xlabel('Count')
plt.ylabel('Continent')
plt.show()
```

```python
# 3. Coalition Memberships (EU, OECD, G20, G7, ASEAN)
alliance_cols = ['eu', 'oecd', 'g20', 'g7', 'asean']
alliance_counts = df_cleaned[alliance_cols].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x=alliance_counts.values, y=alliance_counts.index, palette='magma')
plt.title('Count of Entities in Each Coalition/Alliance')
plt.xlabel('Number of Entities')
plt.ylabel('Coalition / Alliance')
plt.show()
```

### Summary of Country Memberships in Alliances


```python
# Filter for countries/economies
countries_df = df_cleaned[df_cleaned['area_type'] == 'Country or economy']

# Calculate sums of membership flags
alliance_columns = ['eu', 'oecd', 'g20', 'g7', 'asean']
alliance_country_counts = countries_df[alliance_columns].sum().reset_index()
alliance_country_counts.columns = ['Alliance', 'Country Count']

# Sort in descending order
alliance_country_counts = alliance_country_counts.sort_values(by='Country Count', ascending=False).reset_index(drop=True)

display(alliance_country_counts)
```

### Visualizing Alliance Membership Distribution with a Pie Chart


```python
plt.figure(figsize=(8, 8))
plt.pie(
    alliance_country_counts['Country Count'],
    labels=alliance_country_counts['Alliance'].str.upper(),
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('pastel')[0:5]
)
plt.title('Proportion of Alliance Memberships among Countries/Economies')
plt.show()
```
